## 12. Loops

Loops repeat code.

A `for` loop is often used when we want to go through a list.

In Deep Learning, loops are the backbone of our **training loop**.
- An **epoch** is one full iteration over the data.
- We process data in smaller chunks called **mini-batches**.

You will often use `range()` to loop a specific number of times (e.g., number of epochs), and `enumerate()` to get both the index (batch number) and the item itself.

````python
_loss_values = [0.92, 0.74, 0.61, 0.50]

# Standard loop
for _loss in _loss_values:
    print("Loss:", _loss)

print("---")

# Using enumerate to track the batch or _epoch index
for epoch_idx, _loss in enumerate(_loss_values):
    print(f"Epoch {epoch_idx + 1} - Loss: {_loss}")

print("---")

# Using range for a set number of iterations
for i in range(3):
    print("Training step:", i)
````


A `while` loop repeats as long as a condition stays true.


````python
_epoch = 1

while _epoch <= 3:
    print("Training _epoch:", _epoch)
    _epoch += 1
````

## 15. Objects and classes

In Python, objects are created from **classes**.

Key concepts:
- **Class** → blueprint for creating objects
- **Object** → instance of a class
- **Attribute** → data stored in an object
- **Method** → function defined inside a class
### Why do we care in Deep Learning?
In PyTorch, **every Neural Network is an object**. 
- The **parameters** (weights and biases) of the network are stored as its **attributes**.
- The **forward pass** (the math that turns inputs into predictions) is implemented as a **method** (typically called `forward` or via `__call__`).

````python
# Define a class
class Dog:
    def __init__(self, name, age):
        # Attributes (data)
        self.name = name
        self.age = age

    # Method (behavior)
    def bark(self):
        print("Woof!")

    def describe(self):
        print(f"{self.name} is {self.age} years old.")


# Create objects (instances of the class)
dog1 = Dog("Tovi", 3)
dog2 = Dog("Luna", 5)

# Access attributes
print(dog1.name)
print(dog2.age)

# Call methods
dog1.bark()
dog2.describe()
````

## 16. Class Inheritance

Classes can **inherit** behavior from other classes. In PyTorch, every single custom Neural Network you write **must** inherit from PyTorch's base class called `torch.nn.Module`. 

To initialize the parent class properly, we use `super().__init__()`. You will copy-paste this practically every time you build a PyTorch model!

````python
class Animal:
    def __init__(self):
        self.is_alive = True

# Dog inherits from Animal
class SpecializedDog(Animal):
    def __init__(self, name):
        # Initialize the parent class (Animal) to get 'is_alive'
        super().__init__() 
        self.name = name
        
    def bark(self):
        print(f"{self.name} says Woof!")

my_dog = SpecializedDog("Luna")
print("Name:", my_dog.name)
print("Is alive?:", my_dog.is_alive)
my_dog.bark()
````

## 17. Context Managers & Decorators

In PyTorch, you will often see keywords like `with` or the `@` symbol. These are advanced Python features that are incredibly useful in Deep Learning.

### The `with` statement (Context Managers)
The `with` statement is used to wrap a block of code with setup and teardown actions. It guarantees that certain resources are properly cleaned up or conditions are met, even if an error occurs. In standard Python, it is commonly used to open and read files without manually having to close them.

In PyTorch, you will constantly use `with torch.no_grad():` to temporarily disable gradient tracking during model evaluation, saving memory and computation.

````python
# Example 1: File handling (built-in Python)
with open("test.txt", "w") as f:
    f.write("Hello Deep Learning!")
    # The file automatically closes when we leave this indented block!

# Example 2: PyTorch context manager (concept)
# This temporarily tells PyTorch to stop tracking operations for gradients.
import torch
with torch.no_grad():
    # Imagine passing images through our model for testing
    predictions = "Model inference without tracking gradients"
    print(predictions)
````

### Decorators (`@`)
A **decorator** is a function that modifies the behavior of another function. It is placed on the line immediately preceding the function definition using the `@` symbol.

In PyTorch, `@torch.no_grad()` is an alternative to `with torch.no_grad():`. By putting it above a function, the whole function is safely run without tracking gradients.

````python
# This is a basic example of applying a decorator (concept)
@torch.no_grad()
def evaluate_model():
    print("Evaluating without gradients safely!")

evaluate_model()
````

## 18. Error Handling (try/except)

Python allows you to catch and handle errors to prevent your entire program from crashing. In Deep Learning, you might want to catch `RuntimeError` (like Out-Of-Memory/OOM exceptions) or catch missing files when building your datasets.

````python
try:
    # Code that might fail
    calculation = 10 / 0
    print("This won't be printed.")
except ZeroDivisionError:
    # What to do if it fails
    print("Error: You tried to divide by zero! Safely caught the exception.")
except Exception as e:
    # Catching any other general error
    print(f"An unexpected error occurred: {e}")
````
