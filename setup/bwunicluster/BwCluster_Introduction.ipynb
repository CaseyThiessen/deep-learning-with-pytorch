## 5. Setup a Python virtual environment

A virtual environment keeps your project packages separate from the system Python installation.

### 5.1 Create the virtual environment
Open a terminal either:

- Inside the Jupyter browser interface, 
- Inside the graphical SHH client interface,
- Through SSH

Then move into the course repository root (the directory containing [requirements.txt](../../requirements.txt)) and create an environment by running the following commands one after the other:


````python
# Cluster terminal or Jupyter terminal
cd /path/to/deep-learning-with-pytorch

python3.12 -m venv deep_learning_venv

source deep_learning_venv/bin/activate

pip install --upgrade pip

pip install torch==2.11.0 torchvision==0.26.0 torchaudio==2.11.0 --index-url https://download.pytorch.org/whl/cu128

pip install -r requirements.txt

python -m ipykernel install \
    --user \
    --name Deep_Learning_with_Pytorch \
    --display-name "Deep_Learning_with_Pytorch"

deactivate
````

### 5.2 Fix the Jupyter notebook bug
The bug fix is needed because Jupyter’s default kernel configuration may point directly to the virtual environment’s Python executable, which can fail on the cluster when the environment is not initialized correctly. Replacing the kernel’s argv entry with a clean startup script ensures that the virtual environment is activated properly before Jupyter launches the kernel, so notebooks use the right packages, paths, and runtime settings.

The [clean kernel wrapper](clean_kernel.sh) is already included at `setup/bwunicluster/clean_kernel.sh`.

Open that file and replace its Python-path placeholder with the absolute path to the environment created above, for example `/path/to/deep-learning-with-pytorch/deep_learning_venv/bin/python`.

Then use the following code to access the faulty kernel file:

````python
# Cluster terminal or Jupyter terminal
nano ~/.local/share/jupyter/kernels/deep_learning_with_pytorch/kernel.json
````

Replace only the `argv` value in your existing `kernel.json` with the following list, using the absolute path to the wrapper in your cloned repository. Leave the other kernel settings unchanged:

```json
[
  "/path/to/deep-learning-with-pytorch/setup/bwunicluster/clean_kernel.sh",
  "-f",
  "{connection_file}"
]
```

### 6.1 Slurm commands

Slurm is an open source, fault-tolerant, and highly scalable cluster management and job scheduling system for large and small Linux clusters.

Here are the most common commands (used in the cluster terminal).

| Slurm commands    | Brief explanation                                                                    |
|-------------------|--------------------------------------------------------------------------------------|
| sbatch            | Submits a job and puts it into the queue                                             |
| salloc            | Requests resources for an interactive Job                                            |
| scontrol show job | Displays detailed job state information                                              |
| squeue            | Displays information about active, eligible, blocked, and/or recently completed jobs |
| squeue --start    | Returns start time of submitted job                                                  |
| sinfo_t_idle      | Shows what resources are available for immediate use                                 |
| scancel           | Cancels a job                                                                        |

The most important commands you will be working with are:
- sbatch 
- squeue 
- sinfo_t_idle

### 6.2 Partitions

A partition is a queue or group of cluster nodes with similar hardware and usage rules. In this case, the relevant partitions provide access to different types of GPUs, but the partition itself is not the GPU; it determines which group of machines your job may run on. On the cluster, we can choose from a variety of GPU partitions for our computations. Some partitions are intended for short testing and debugging jobs, while others are meant for longer experiments.

For testing and debugging, we recommend using the `dev_gpu_h100 partition`. It is more easily available than other partitions, but has a maximum runtime of 30 minutes.

For the actual experiments of your project that have a longer computation time than 30 minutes, we recommend using `gpu_a100_il`, `gpu_h100_il`, or `gpu_h100`.

### 6.3 Resources, Job File, and Job Submission

A job is a task that you submit to the cluster, such as running a script or program, together with instructions about the resources it needs. Resources required to run your job can be things, such as memory, runtime, number of tasks, or GPU resources. Batch jobs on the BwCluster are submitted using the sbatch command. After submission, Slurm places the job in a queue. The job starts once the requested resources are available and depending on the current fair-share scheduling priorities.

You can specify the required resources directly in the terminal. For example, to submit a serial job that runs `setup/bwunicluster/job.sh`, requests 5000 MB of memory, and has a wall-clock time limit of 10 minutes, run from the repository root:

````python
sbatch -p dev_cpu -n 1 -t 10:00 --mem=5000 --job-name Proof_of_Concept setup/bwunicluster/job.sh
````

However, we recommend defining the required resources and additional options directly inside `setup/bwunicluster/job.sh`. This makes the job configuration easier to read, modify, and reuse.

A typical job.sh file may contain the following #SBATCH options:

| sbatch option                      | Description                                                                                                                  |
|------------------------------------|------------------------------------------------------------------------------------------------------------------------------|
| `#SBATCH --partition=queue-name`   | Request a specific queue for the resource allocation.                                                                        |
| `#SBATCH --ntasks=count`           | Number of tasks to be launched.                                                                                              |
| `#SBATCH --time=time`              | Wall clock time limit.                                                                                                       |
| `#SBATCH --job-name=name`          | Job name.                                                                                                                    |
| `#SBATCH --gres=gpu:count`         | Request GPU resources for the job.                                                                                           |
| `#SBATCH --output=name`            | File in which job output is stored.                                                                                          |
| `#SBATCH --error=name`             | File in which job error messages are stored.                                                                                 |
| `#SBATCH --mail-type=type`         | Notify user by email when certain event types occur. Valid type values are `NONE`, `BEGIN`, `END`, `FAIL`, `REQUEUE`, `ALL`. |
| `#SBATCH --mail-user=mail-address` | The specified email address receives email notification of state changes as defined by `--mail-type`.                        |

These settings can be enabled or disabled by adding or removing an additional # at the beginning of the corresponding line. 

For debugging and testing, we recommend disabling email notifications to avoid unnecessary messages.

More customization options are available here:
https://wiki.bwhpc.de/e/BwUniCluster3.0/Running_Jobs/Slurm

The final line of your job.sh file usually specifies which program or Python script should be executed on the cluster.

After configuring the job script, submit the job from the repository root in the cluster terminal with:

````python
# Cluster terminal 
sbatch setup/bwunicluster/job.sh
````

Slurm will then queue the job and run it once the requested resources become available.

## 7. Troubleshooting checklist

### I cannot log in

- Check that you are using the correct username.
- Check whether you are connected to the university network or VPN.
- Check whether your two-factor authentication code is current.

### SSH works, but file transfer does not

- Try SFTP instead of plain FTP.
- On Windows, try WinSCP if MobaXterm gives problems.
- In MobaXterm 23.6, switch the SSH engine to `<legacy>` and restart MobaXterm.

### Jupyter does not show my environment

- Make sure you ran `python -m ipykernel install ...` while the environment was active.
- Refresh the browser.
- Restart the Jupyter session if needed.

### My package import fails

- Check that you selected the correct kernel.
- Activate the environment and run `pip list`.
- Install the missing package inside the same environment.

### I cannot find my files

- Run `pwd` in the terminal to check where you are.
- Make sure you copied files to the cluster, not only to your local computer.
